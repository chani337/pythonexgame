# 파이썬 함수 정리

## 1. 함수란?

함수는 **자주 사용하는 코드를 하나로 묶어두고 필요할 때마다 불러서 사용하는 기능**입니다.  
쉽게 말하면 **코드 상자** 또는 **기능 묶음**이라고 생각하면 됩니다.

예를 들어,

- 인사말 출력하기
- 두 수 더하기
- 점수 평균 구하기

같은 기능을 함수로 만들 수 있습니다.

---

## 2. 함수의 기본 형태

```python
def 함수이름():
    실행할 코드
```

### 예시

```python
def hello():
    print("안녕하세요")

hello()
```

### 실행 결과

```python
안녕하세요
```

---

## 3. 함수는 왜 사용할까?

함수를 사용하면

- 같은 코드를 여러 번 쓰지 않아도 되고
- 코드를 더 읽기 쉽게 만들 수 있고
- 수정이 쉬워집니다

즉, **코드를 더 효율적으로 관리할 수 있습니다.**

---

## 4. 매개변수가 있는 함수

함수는 값을 받아서 사용할 수도 있습니다.  
이때 함수 괄호 안에 들어가는 변수를 **매개변수**라고 합니다.

```python
def hello(name):
    print(name + "님, 안녕하세요")

hello("홍길동")
```

### 실행 결과

```python
홍길동님, 안녕하세요
```

---

## 5. 값을 반환하는 함수

함수는 결과값을 돌려줄 수도 있습니다.  
이때 `return`을 사용합니다.

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
```

### 실행 결과

```python
8
```

## 7. 함수 실습 예제 2: 두 수 더하기

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

---

## 8. 함수 실습 예제 3: 제곱값 반환하기

```python
def square(num):
    return num * num

print(square(4))
```

### 실행 결과

```python
16
```

함수를 만들기만 하고 부르지 않으면 실행되지 않습니다.

## 10. 함수와 입력값

```python
def introduce(name, age):
    print("이름은", name, "이고 나이는", age, "살입니다.")

introduce("김철수", 20)
```

## 11. 자주 하는 실수

### 1) 함수만 만들고 호출하지 않는 경우

```python
def hello():
    print("안녕하세요")
```

이렇게만 쓰면 출력되지 않습니다.  
아래처럼 호출해야 합니다.

```python
hello()
```

### 2) `return`과 `print`를 헷갈리는 경우

```python
def add(a, b):
    print(a + b)
```

이 함수는 값을 출력만 하고 돌려주지는 않습니다.  
결과를 나중에 저장해서 사용하려면 `return`이 필요합니다.

```python
def add(a, b):
    return a + b
```

### 3) 들여쓰기를 틀리는 경우

```python
def hello():
print("안녕하세요")
```

함수 안의 코드는 반드시 들여쓰기 해야 합니다.

## 12. 한 번에 정리하는 함수 예제

```python
# 함수 정의
def add(a, b):
    return a + b

# 함수 호출
result1 = add(3, 4)
result2 = add(10, 20)

print(result1)
print(result2)
```

---

## 13. 핵심 정리

- 함수는 코드를 묶어두고 필요할 때 사용하는 기능
- `def`로 함수를 정의함
- 함수는 호출해야 실행됨
- 함수는 매개변수를 받을 수 있음
- `return`으로 결과값을 돌려줄 수 있음
- 함수 안의 코드는 들여쓰기가 중요함
