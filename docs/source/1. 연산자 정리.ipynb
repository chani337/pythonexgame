# 파이썬 연산자 정리

## 1. 연산자란?

연산자는 **값을 계산하거나, 비교하거나, 조건을 판단할 때 사용하는 기호**입니다.

쉽게 말하면 **숫자를 계산하거나, 값을 비교하거나, 조건이 맞는지 확인할 때 사용하는 표시**입니다.

예를 들어,

- `3 + 2` → 더하기
- `10 > 5` → 크기 비교
- `a == b` → 같은지 확인

처럼 다양한 상황에서 사용합니다.

---

## 2. 산술 연산자

산술 연산자는 **숫자를 계산할 때** 사용합니다.

| 연산자 | 의미 | 예시 | 결과 |
|---|---|---|---:|
| `+` | 더하기 | `3 + 2` | `5` |
| `-` | 빼기 | `5 - 2` | `3` |
| `*` | 곱하기 | `4 * 3` | `12` |
| `/` | 나누기 | `10 / 2` | `5.0` |
| `//` | 몫 | `10 // 3` | `3` |
| `%` | 나머지 | `10 % 3` | `1` |
| `**` | 거듭제곱 | `2 ** 3` | `8` |

### 예시

```python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 3)
```

### 실행 결과

```text
13
7
30
3.3333333333333335
3
1
1000
```

### 핵심

- `/` → 나눈 결과를 계산하며 **실수(float)** 로 반환
- `//` → **몫**을 구함
- `%` → **나머지**를 구함
- `**` → **거듭제곱**을 계산

---

## 3. 비교 연산자

비교 연산자는 **두 값을 비교하여 `True` 또는 `False`로 판단**합니다.

| 연산자 | 의미 | 예시 | 결과 |
|---|---|---|---|
| `==` | 같다 | `3 == 3` | `True` |
| `!=` | 다르다 | `3 != 2` | `True` |
| `>` | 크다 | `5 > 3` | `True` |
| `<` | 작다 | `2 < 4` | `True` |
| `>=` | 크거나 같다 | `5 >= 5` | `True` |
| `<=` | 작거나 같다 | `3 <= 2` | `False` |

### 예시

```python
print(5 == 5)
print(5 != 3)
print(10 > 7)
print(4 < 2)
print(6 >= 6)
print(3 <= 1)
```

### 실행 결과

```text
True
True
True
False
True
False
```

### 핵심

비교 연산자의 결과는 기본적으로 **`True` 또는 `False`**입니다.

---

## 4. 대입 연산자

대입 연산자는 **변수에 값을 저장하거나 기존 값을 변경할 때** 사용합니다.

| 연산자 | 의미 | 예시 |
|---|---|---|
| `=` | 값을 저장 | `x = 10` |
| `+=` | 더한 후 저장 | `x += 3` |
| `-=` | 뺀 후 저장 | `x -= 2` |
| `*=` | 곱한 후 저장 | `x *= 2` |
| `/=` | 나눈 후 저장 | `x /= 2` |
| `//=` | 몫을 구한 후 저장 | `x //= 2` |
| `%=` | 나머지를 구한 후 저장 | `x %= 2` |
| `**=` | 거듭제곱한 후 저장 | `x **= 2` |

### 예시

```python
x = 10

x += 5
print(x)

x -= 3
print(x)

x *= 2
print(x)
```

### 실행 결과

```text
15
12
24
```

예를 들어,

```python
x += 5
```

는 다음과 같은 의미입니다.

```python
x = x + 5
```

## 5. 논리 연산자

논리 연산자는 **여러 조건을 함께 판단할 때** 사용합니다.

| 연산자 | 의미 | 설명 |
|---|---|---|
| `and` | 그리고 | 두 조건이 모두 `True`여야 `True` |
| `or` | 또는 | 하나라도 `True`이면 `True` |
| `not` | 아니다 | `True`와 `False`를 반대로 바꿈 |

### 예시

```python
print(10 > 5 and 3 > 1)
print(10 > 5 and 3 < 1)
print(10 > 5 or 3 < 1)
print(not (10 > 5))
```

### 실행 결과

```text
True
False
True
False
```

### 쉽게 기억하기

```text
and → 둘 다 참이어야 참
or  → 하나만 참이어도 참
not → 참과 거짓을 반대로
```

## 6. 문자열에서도 연산자를 사용할 수 있다

연산자는 숫자뿐만 아니라 **문자열에서도 일부 사용할 수 있습니다.**

### 문자열 이어 붙이기 `+`

문자열에 `+`를 사용하면 문자열을 이어 붙일 수 있습니다.

```python
text1 = "안녕"
text2 = "하세요"

print(text1 + text2)
```

### 실행 결과

```text
안녕하세요
```

### 문자열 반복하기 `*`

문자열에 숫자를 곱하면 문자열을 반복할 수 있습니다.

```python
print("파이썬" * 3)
```

### 실행 결과

```text
파이썬파이썬파이썬
```

## 7. 연산자 우선순위

여러 연산자를 한꺼번에 사용하면 **정해진 순서에 따라 계산**됩니다.

기본적인 순서는 다음과 같이 이해하면 됩니다.

1. 괄호 `()`
2. 거듭제곱 `**`
3. 곱하기, 나누기, 몫, 나머지 `*`, `/`, `//`, `%`
4. 더하기, 빼기 `+`, `-`
5. 비교 연산자 `>`, `<`, `>=`, `<=`, `==`, `!=`
6. 논리 연산자 `not`, `and`, `or`
7. 대입 연산자 `=`, `+=`, `-=`, `*=`, `/=` 등

### 예시

```python
print(3 + 2 * 4)
print((3 + 2) * 4)
```

### 실행 결과

```text
11
20
```

첫 번째는 곱셈이 먼저 계산됩니다.

```text
3 + 2 * 4
→ 3 + 8
→ 11
```

두 번째는 괄호 안의 계산이 먼저 이루어집니다.

```text
(3 + 2) * 4
→ 5 * 4
→ 20
```

따라서 **계산 순서를 명확하게 하고 싶다면 괄호를 사용하는 것이 좋습니다.**

## 8. 삼항 연산자

삼항 연산자는 **조건에 따라 두 가지 값 중 하나를 선택할 때** 사용하는 문법입니다.

파이썬에서는 다음과 같은 형태로 작성합니다.

```python
참일 때의 값 if 조건 else 거짓일 때의 값
```

쉽게 말하면,

> **조건이 참이면 앞의 값을 선택하고, 거짓이면 뒤의 값을 선택합니다.**

### 기본 형태

```python
값1 if 조건 else 값2
```

- 조건이 `True` → `값1`
- 조건이 `False` → `값2`

### 예시

```python
age = 20

result = "성인" if age >= 18 else "미성년자"

print(result)
```

### 실행 결과

```text
성인
```

`age >= 18`이 `True`이므로 `"성인"`이 선택됩니다.

### 예시 2: 숫자 비교

```python
a = 10
b = 5

result = a if a > b else b

print(result)
```

### 실행 결과

```text
10
```

`a > b`가 `True`이므로 `a`의 값인 `10`이 선택됩니다.

### 예시 3: 홀수와 짝수 판단

```python
number = 7

result = "짝수" if number % 2 == 0 else "홀수"

print(result)
```

### 실행 결과

```text
홀수
```

`number % 2 == 0`이 `False`이므로 `"홀수"`가 선택됩니다.

### 일반적인 if-else문과 비교

삼항 연산자를 사용하지 않으면 다음과 같이 작성할 수 있습니다.

```python
age = 20

if age >= 18:
    result = "성인"
else:
    result = "미성년자"

print(result)
```

삼항 연산자를 사용하면 다음처럼 짧게 표현할 수 있습니다.

```python
age = 20

result = "성인" if age >= 18 else "미성년자"

print(result)
```

두 코드는 같은 결과를 냅니다.

### 주의할 점

다른 프로그래밍 언어에서는 다음과 같은 형태의 삼항 연산자를 사용하는 경우가 있습니다.

```text
조건 ? 참일 때 값 : 거짓일 때 값
```

하지만 **파이썬에서는 `?`를 사용하지 않습니다.**

파이썬에서는 다음과 같이 작성합니다.

```python
참일 때 값 if 조건 else 거짓일 때 값
```

삼항 연산자는 **간단한 조건을 한 줄로 표현할 때** 유용합니다.

조건이 복잡해지면 일반적인 `if-else`문을 사용하는 것이 더 읽기 쉽습니다.

## 13. 자주 하는 실수

### 1) `=`와 `==`를 헷갈리는 경우

```python
x = 10
print(x == 10)
```

- `=` → 값을 저장하는 **대입 연산자**
- `==` → 같은지 비교하는 **비교 연산자**

즉,

```python
x = 10
```

은 `x`에 `10`을 저장하는 것이고,

```python
x == 10
```

은 `x`가 `10`과 같은지 확인하는 것입니다.

### 2) `/`와 `//`를 헷갈리는 경우

```python
print(7 / 2)
print(7 // 2)
```

### 실행 결과

```text
3.5
3
```

- `/` → 나눈 결과
- `//` → 몫

입니다.

### 3) `and`와 `or`를 헷갈리는 경우

```python
print(True and False)
print(True or False)
```

### 실행 결과

```text
False
True
```

- `and` → 둘 다 `True`여야 `True`
- `or` → 하나라도 `True`이면 `True`

입니다.

### 4) 삼항 연산자의 순서를 헷갈리는 경우

파이썬의 삼항 연산자는 다음 순서로 작성합니다.

```python
참일 때 값 if 조건 else 거짓일 때 값
```

다음과 같은 형태는 파이썬에서 사용하지 않습니다.

```text
조건 ? 참일 때 값 : 거짓일 때 값
```

## 15. 핵심 정리

| 종류 | 주요 연산자 | 용도 |
|---|---|---|
| 산술 연산자 | `+`, `-`, `*`, `/`, `//`, `%`, `**` | 숫자 계산 |
| 비교 연산자 | `==`, `!=`, `>`, `<`, `>=`, `<=` | 값 비교 |
| 대입 연산자 | `=`, `+=`, `-=`, `*=`, `/=` | 변수에 값 저장 및 변경 |
| 논리 연산자 | `and`, `or`, `not` | 여러 조건 판단 |
| 문자열 연산 | `+`, `*` | 문자열 연결 및 반복 |
| 삼항 연산자 | `값1 if 조건 else 값2` | 조건에 따라 값 선택 |

### 꼭 기억할 것

- `+`, `-`, `*`, `/` → 기본적인 계산
- `//` → 몫
- `%` → 나머지
- `**` → 거듭제곱
- `==` → 같은지 비교
- `=` → 값을 저장
- `and` → 모두 참이어야 참
- `or` → 하나라도 참이면 참
- `not` → 참과 거짓을 반대로
- `if ... else ...` → 조건에 따라 값 선택

## 16. 마무리

연산자는 파이썬에서 가장 기본적이면서도 매우 중요한 문법입니다.

앞으로 배우게 될

- 조건문
- 반복문
- 리스트와 딕셔너리
- 함수
- 데이터 분석

등에서도 연산자를 계속 사용하게 됩니다.

특히 다음 네 가지는 꼭 익혀두는 것이 좋습니다.

```python
# 계산
10 + 3

# 비교
10 > 5

# 조건 판단
10 > 5 and 3 > 1

# 조건에 따라 값 선택
"성인" if age >= 18 else "미성년자"
```

연산자의 종류와 사용 방법을 익혀두면 이후의 파이썬 문법을 훨씬 쉽게 이해할 수 있습니다.
