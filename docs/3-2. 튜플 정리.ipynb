# 파이썬 튜플 정리

## 1. 튜플이란?

튜플은 **여러 개의 값을 하나로 묶어서 저장하는 자료형**입니다.  
리스트와 비슷하지만, 한 번 만들고 나면 **값을 수정할 수 없다는 특징**이 있습니다.

쉽게 말하면 튜플은 **바꿀 수 없는 리스트**라고 생각하면 됩니다.

예를 들어,

- 좌표 값 저장하기
- 색상 정보 저장하기
- 여러 개의 값을 한 번에 반환하기

같은 경우에 사용할 수 있습니다.

파이썬에서는 소괄호 `()`를 사용하여 튜플을 만듭니다.

---

## 2. 튜플의 기본 형태

```python
튜플이름 = (값1, 값2, 값3)
```

### 예시

```python
numbers = (10, 20, 30)
print(numbers)
```

### 실행 결과

```python
(10, 20, 30)
```

---

## 3. 튜플의 위치 번호(인덱스)

튜플도 리스트처럼 **순서가 있는 자료형**이기 때문에  
인덱스를 이용해서 값을 꺼낼 수 있습니다.

주의할 점은 **첫 번째 값의 인덱스는 0부터 시작**한다는 것입니다.

```python
fruits = ("사과", "바나나", "포도")

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

### 실행 결과

```python
사과
바나나
포도
```

---

## 4. 튜플은 값 수정이 불가능하다

튜플은 한 번 만들면 값을 바꿀 수 없습니다.

```python
fruits = ("사과", "바나나", "포도")
fruits[1] = "수박"
```

위 코드는 오류가 납니다.

즉, 튜플은 **값 추가, 수정, 삭제가 불가능한 자료형**입니다.

---

## 5. 값이 하나인 튜플 만들기

값이 하나만 들어 있는 튜플은 **쉼표를 반드시 함께 써야 합니다.**

```python
x = (10,)
print(x)
```

### 실행 결과

```python
(10,)
```

쉼표가 없으면 그냥 숫자나 문자열로 인식될 수 있습니다.

```python
x = (10)
print(x)
```

### 실행 결과

```python
10
```

---

## 6. 튜플의 길이 구하기

튜플에 값이 몇 개 들어 있는지 알고 싶을 때는 `len()`을 사용합니다.

```python
numbers = (10, 20, 30, 40)

print(len(numbers))
```

### 실행 결과

```python
4
```

## 10. 리스트와 튜플의 차이

리스트와 튜플은 비슷하지만 중요한 차이가 있습니다.

| 구분 | 리스트 | 튜플 |
|---|---|---|
| 기호 | `[]` | `()` |
| 값 수정 | 가능 | 불가능 |
| 값 추가/삭제 | 가능 | 불가능 |

예를 들어,

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
```

- `my_list`는 수정 가능
- `my_tuple`은 수정 불가능

---

## 11. 자주 하는 실수

### 1) 튜플도 리스트처럼 수정하려는 경우

```python
numbers = (10, 20, 30)
numbers[1] = 99
```

튜플은 값을 바꿀 수 없기 때문에 오류가 납니다.

### 2) 값 하나짜리 튜플에서 쉼표를 빼먹는 경우

```python
x = (10)
print(type(x))
```

이 경우 튜플이 아니라 정수로 인식됩니다.

```python
x = (10,)
print(type(x))
```

이렇게 써야 튜플이 됩니다.

### 3) 괄호만 보고 리스트와 헷갈리는 경우

- 리스트는 `[]`
- 튜플은 `()`

기호가 다르므로 구분해서 써야 합니다.

## 12. 한 번에 정리하는 튜플 예제

```python
# 튜플 생성
student = ("홍길동", 20, "인공지능")

# 값 출력
print(student[0])
print(student[1])
print(student[2])

# 길이 출력
print(len(student))
```

### 실행 결과

```python
홍길동
20
인공지능
3
```

---

## 13. 핵심 정리

- 튜플은 여러 값을 한 번에 저장하는 자료형
- 소괄호 `()`를 사용함
- 인덱스는 0부터 시작함
- 값 조회는 가능함
- 값 수정, 추가, 삭제는 불가능함
- 값이 하나인 튜플은 쉼표가 반드시 필요함

---

## 14. 마무리

튜플은 앞으로 배우게 될

- 여러 값 묶음 처리
- 함수의 반환값
- 좌표 데이터 저장
- 수정되면 안 되는 데이터 관리

에서 자주 사용됩니다.  
특히 **변경되면 안 되는 값들을 안전하게 저장할 때** 유용합니다.
