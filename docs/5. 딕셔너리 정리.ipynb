# 파이썬 딕셔너리 정리

## 1. 딕셔너리란?

딕셔너리는 **키(key)와 값(value)을 한 쌍으로 저장하는 자료형**입니다.  
쉽게 말하면 **단어장처럼 이름표와 내용이 함께 저장되는 구조**입니다.

예를 들어,

- 이름 : 홍길동
- 나이 : 20
- 학과 : 인공지능

처럼 **항목 이름과 값**을 같이 저장할 때 사용합니다.

파이썬에서는 중괄호 `{}`를 사용합니다.

---

## 2. 딕셔너리의 기본 형태

```python
딕셔너리이름 = {키1: 값1, 키2: 값2, 키3: 값3}
```

### 예시

```python
student = {"이름": "홍길동", "나이": 20, "학과": "인공지능"}

print(student)
```

### 실행 결과

```python
{'이름': '홍길동', '나이': 20, '학과': '인공지능'}
```

---

## 3. 딕셔너리 값 꺼내기

딕셔너리의 값은 **키를 이용해서** 꺼냅니다.

```python
student = {"이름": "홍길동", "나이": 20, "학과": "인공지능"}

print(student["이름"])
print(student["나이"])
```

### 실행 결과

```python
홍길동
20
```

---

## 4. 딕셔너리 값 수정하기

기존 키에 새로운 값을 넣으면 값이 수정됩니다.

```python
student = {"이름": "홍길동", "나이": 20}
student["나이"] = 21

print(student)
```

### 실행 결과

```python
{'이름': '홍길동', '나이': 21}
```

---

## 5. 딕셔너리에 값 추가하기

없는 키에 값을 넣으면 새로운 항목이 추가됩니다.

```python
student = {"이름": "홍길동", "나이": 20}
student["학과"] = "인공지능"

print(student)
```

### 실행 결과

```python
{'이름': '홍길동', '나이': 20, '학과': '인공지능'}
```

---

## 6. 딕셔너리에서 값 삭제하기

`del`을 사용하면 특정 키와 값을 삭제할 수 있습니다.

```python
student = {"이름": "홍길동", "나이": 20, "학과": "인공지능"}

del student["나이"]

print(student)
```

### 실행 결과

```python
{'이름': '홍길동', '학과': '인공지능'}
```

---

## 7. 딕셔너리의 키와 값 확인하기

```python
student = {"이름": "홍길동", "나이": 20, "학과": "인공지능"}

print(student.keys())
print(student.values())
```

## 11. 자주 하는 실수

### 1) 인덱스로 값을 꺼내려고 하는 경우

```python
student = {"이름": "홍길동", "나이": 20}
print(student[0])
```

딕셔너리는 리스트처럼 번호로 꺼내는 것이 아니라 **키로 꺼내야 합니다**.

```python
print(student["이름"])
```

### 2) 없는 키를 사용하는 경우

```python
student = {"이름": "홍길동", "나이": 20}
print(student["학년"])
```

없는 키를 사용하면 오류가 납니다.

### 3) 키와 값을 구분하지 못하는 경우

딕셔너리는 **키 : 값** 구조입니다.

```python
student = {"이름": "홍길동"}
```

여기서 `"이름"`은 키이고, `"홍길동"`은 값입니다.

## 12. 한 번에 정리하는 딕셔너리 예제

```python
# 딕셔너리 생성
student = {"이름": "홍길동", "나이": 20}

# 값 출력
print(student["이름"])

# 값 수정
student["나이"] = 21

# 값 추가
student["학과"] = "인공지능"

# 전체 출력
print(student)
```

---

## 13. 핵심 정리

- 딕셔너리는 키와 값을 한 쌍으로 저장하는 자료형
- 중괄호 `{}`를 사용함
- 값은 인덱스가 아니라 키로 꺼냄
- 기존 키에 값을 넣으면 수정됨
- 없는 키에 값을 넣으면 추가됨
- `del`로 삭제 가능
