# 파이썬 리스트 정리

## 1. 리스트란?

리스트는 **여러 개의 값을 하나의 변수에 묶어서 저장하는 자료형**입니다.  
쉽게 말하면 **값들을 순서대로 담아두는 바구니**라고 생각하면 됩니다.

예를 들어,

- 학생 이름 여러 개
- 점수 여러 개
- 과일 이름 여러 개

처럼 여러 데이터를 한 번에 관리할 때 사용합니다.

파이썬에서는 대괄호 `[]`를 사용하여 리스트를 만듭니다.

---

## 2. 리스트의 기본 형태

```python
리스트이름 = [값1, 값2, 값3]
```

### 예시

```python
fruits = ["사과", "바나나", "포도"]
print(fruits)
```

### 실행 결과

```python
['사과', '바나나', '포도']
```

---

## 3. 리스트의 위치 번호(인덱스)

리스트 안에 있는 값은 **순서 번호**로 꺼낼 수 있습니다.  
이 번호를 **인덱스(index)** 라고 합니다.

주의할 점은 **첫 번째 값의 인덱스는 0부터 시작**한다는 것입니다.

```python
fruits = ["사과", "바나나", "포도"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

### 실행 결과

```python
사과
바나나
포도
```

---

## 4. 리스트 값 수정하기

리스트는 저장된 값을 나중에 바꿀 수 있습니다.

```python
fruits = ["사과", "바나나", "포도"]
fruits[1] = "수박"

print(fruits)
```

### 실행 결과

```python
['사과', '수박', '포도']
```

---

## 5. 리스트에 값 추가하기

리스트에 새로운 값을 추가할 때는 `append()`를 사용합니다.

```python
fruits = ["사과", "바나나"]
fruits.append("포도")

print(fruits)
```

### 실행 결과

```python
['사과', '바나나', '포도']
```

---

## 6. 리스트에서 값 삭제하기

리스트에서 값을 삭제할 때는 `remove()`를 사용할 수 있습니다.

```python
fruits = ["사과", "바나나", "포도"]
fruits.remove("바나나")

print(fruits)
```

### 실행 결과

```python
['사과', '포도']
```

---

## 7. 리스트의 길이 구하기

리스트에 값이 몇 개 들어 있는지 알고 싶을 때는 `len()`을 사용합니다.

```python
fruits = ["사과", "바나나", "포도"]

print(len(fruits))
```

### 실행 결과

```python
3
```

---

## 8. 리스트 실습 예제 1: 점수 저장하기

```python
scores = [80, 90, 75]

print(scores[0])
print(scores[1])
print(scores[2])
```

---

## 9. 리스트 실습 예제 2: 값 추가하기

```python
animals = ["강아지", "고양이"]
animals.append("토끼")

print(animals)
```

---

## 10. 리스트 실습 예제 3: 값 변경하기

```python
colors = ["빨강", "파랑", "노랑"]
colors[2] = "초록"

print(colors)
```

## 11. 자주 하는 실수

### 1) 인덱스를 1부터 생각하는 경우

```python
fruits = ["사과", "바나나", "포도"]
print(fruits[1])
```

이 코드는 `"바나나"`가 출력됩니다.  
파이썬은 **0부터 시작**하기 때문입니다.

### 2) 없는 위치를 꺼내는 경우

```python
fruits = ["사과", "바나나", "포도"]
print(fruits[3])
```

리스트에 3번 위치는 없기 때문에 오류가 납니다.

### 3) 괄호를 잘못 쓰는 경우

```python
fruits = ["사과", "바나나", "포도"]
print(fruits(0))
```

리스트 값은 `()`가 아니라 `[]`로 꺼내야 합니다.

```python
print(fruits[0])
```

## 12. 한 번에 정리하는 리스트 예제

```python
# 과일 리스트 생성
fruits = ["사과", "바나나", "포도"]

# 첫 번째 값 출력
print(fruits[0])

# 값 추가
fruits.append("수박")

# 값 변경
fruits[1] = "딸기"

# 전체 리스트 출력
print(fruits)

# 길이 출력
print(len(fruits))
```

---

## 13. 핵심 정리

- 리스트는 여러 값을 한 번에 저장하는 자료형
- 대괄호 `[]`를 사용함
- 인덱스는 0부터 시작함
- `append()`로 값 추가 가능
- `remove()`로 값 삭제 가능
- `len()`으로 길이 확인 가능
