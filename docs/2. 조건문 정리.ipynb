# 파이썬 조건문 정리

## 1. 조건문이란?

조건문은 **"어떤 조건이 맞으면 이 코드를 실행하고, 아니면 저 코드를 실행하는 문법"** 입니다.  
쉽게 말하면 컴퓨터에게 **상황에 따라 다르게 행동하도록 지시하는 방법**입니다.

예를 들어,

- 비가 오면 우산을 챙긴다.
- 시험 점수가 60점 이상이면 합격이다.
- 나이가 20살 이상이면 성인이다.

이런 판단을 파이썬에서는 `if` 문으로 작성합니다.

---

## 2. 조건문의 기본 형태

```python
if 조건식:
    실행할 코드
```

### 예시

```python
age = 20

if age >= 20:
    print("성인입니다.")
```

### 실행 결과

```python
성인입니다.
```

---

## 3. 들여쓰기 주의

파이썬은 `{ }` 중괄호를 쓰지 않고, **들여쓰기**로 코드의 범위를 구분합니다.

```python
age = 20

if age >= 20:
    print("성인입니다.")
    print("투표가 가능합니다.")
```

`if` 아래에서 **들여쓰기 된 코드들만** 조건이 참일 때 실행됩니다.

---

## 4. 조건식이란?

조건식은 결과가 **참(True)** 또는 **거짓(False)** 으로 나오는 식입니다.

자주 쓰는 비교 연산자:

| 연산자 | 뜻 |
|---|---|
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |
| `==` | 같다 |
| `!=` | 같지 않다 |

### 예시

```python
num = 10

if num > 5:
    print("5보다 큽니다.")
```

---

## 5. if ~ else

조건이 참일 때와 거짓일 때를 나누고 싶으면 `else`를 사용합니다.

```python
score = 75

if score >= 60:
    print("합격입니다.")
else:
    print("불합격입니다.")
```

### 실행 결과

```python
합격입니다.
```

---

## 6. if ~ elif ~ else

조건이 여러 개일 때는 `elif`를 사용합니다.

```python
score = 85

if score >= 90:
    print("A학점")
elif score >= 80:
    print("B학점")
elif score >= 70:
    print("C학점")
else:
    print("재시험 대상")
```

### 실행 결과

```python
B학점
```

---

## 7. 조건문 실습 예제 1: 짝수 / 홀수 판별

```python
num = 7

if num % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
```

### 설명

- `%` 는 나머지를 구하는 연산자입니다.
- 어떤 수를 2로 나눴을 때 나머지가 0이면 짝수입니다.

## 8. 조건문 실습 예제 2: 나이 판별

```python
age = 17

if age >= 20:
    print("성인입니다.")
else:
    print("미성년자입니다.")
```

---

## 9. 조건문 실습 예제 3: 로그인 아이디 확인

```python
user_id = "admin"

if user_id == "admin":
    print("관리자 계정입니다.")
else:
    print("일반 사용자입니다.")
```

## 10. input()과 함께 사용하기

사용자로부터 값을 입력받아 조건문에 사용할 수도 있습니다.

```python
age = int(input("나이를 입력하세요: "))

if age >= 20:
    print("성인입니다.")
else:
    print("미성년자입니다.")
```

### 설명

- `input()`으로 입력받은 값은 기본적으로 **문자열(str)** 입니다.
- 숫자로 계산하거나 비교하려면 `int()`로 감싸서 정수로 변환해야 합니다.

---

## 11. 자주 하는 실수

### 1) `=` 와 `==` 를 헷갈리는 경우

```python
# 잘못된 예
if age = 20:
    print("성인")
```

위 코드는 오류가 납니다.

```python
# 올바른 예
if age == 20:
    print("성인")
```

- `=` : 값을 저장할 때 사용
- `==` : 두 값이 같은지 비교할 때 사용

### 2) 들여쓰기를 안 맞추는 경우

```python
# 잘못된 예
if age >= 20:
print("성인입니다.")
```

파이썬은 들여쓰기가 매우 중요합니다.

```python
# 올바른 예
if age >= 20:
    print("성인입니다.")
```

### 3) input() 값을 바로 숫자처럼 비교하는 경우

```python
age = input("나이를 입력하세요: ")

if age >= 20:
    print("성인입니다.")
```

이 코드는 오류가 날 수 있습니다.  
문자열과 숫자는 바로 비교할 수 없기 때문입니다.

```python
age = int(input("나이를 입력하세요: "))

if age >= 20:
    print("성인입니다.")
```

## 12. 한 번에 정리하는 조건문 예제

```python
# 점수를 입력받아 학점을 출력하는 프로그램

score = int(input("점수를 입력하세요: "))

if score >= 90:
    print("A학점입니다.")
elif score >= 80:
    print("B학점입니다.")
elif score >= 70:
    print("C학점입니다.")
elif score >= 60:
    print("D학점입니다.")
else:
    print("F학점입니다.")
```

---

## 13. 핵심 정리

- `if` : 조건이 참일 때 실행
- `else` : 조건이 거짓일 때 실행
- `elif` : 여러 조건을 순서대로 검사할 때 사용
- 조건식은 `True` 또는 `False` 결과를 가짐
- 파이썬은 **들여쓰기**가 중요함
- `==` 는 비교, `=` 는 저장
