## 2. 변수

변수는 **값에 붙이는 이름**입니다.

```python
name = "찬희"
age = 20

print(name)
print(age)
```

`=`는 같다는 뜻이 아니라 **오른쪽 값을 왼쪽 변수에 저장**한다는 뜻입니다.

### 자료형

| 자료형 | 의미 | 예시 |
|---|---|---|
| `int` | 정수 | `10` |
| `float` | 실수 | `3.14` |
| `str` | 문자열 | `"Python"` |
| `bool` | 참/거짓 | `True`, `False` |

```python
age = 20

print(type(age))
```

파이썬은 저장하는 값에 따라 자료형이 자동으로 결정되는 **동적 자료형 언어**입니다.

```python
x = 10
print(type(x))

x = "Python"
print(type(x))
```

## 3. 변수 값 변경

변수에는 새로운 값을 다시 저장할 수 있습니다.

```python
score = 80
score = 90

print(score)
```

기존 값을 이용해서 변경할 수도 있습니다.

```python
score = 90
score = score + 5

print(score)  # 95
```

## 4. 문자열

문자열은 **글자들의 묶음**이며 따옴표로 감싸서 만듭니다.

```python
name = "Python"
message = "안녕하세요"

print(name)
print(message)
```

### 문자열 연결

`+`를 사용하면 문자열을 이어 붙일 수 있습니다.

```python
first = "Hello"
second = "Python"

print(first + " " + second)
```

문자열을 반복할 때는 `*`를 사용합니다.

```python
print("파이팅! " * 3)
```

## 5. 인덱싱

인덱싱은 문자열에서 **한 글자**를 가져오는 방법입니다.

```python
word = "PYTHON"

print(word[0])   # P
print(word[2])   # T
print(word[-1])  # N
```

- 첫 번째 글자는 `0`
- 두 번째 글자는 `1`
- `-1`은 마지막 글자

```text
P  Y  T  H  O  N
0  1  2  3  4  5

-6 -5 -4 -3 -2 -1
```

---

## 6. 슬라이싱

슬라이싱은 문자열에서 **여러 글자**를 범위로 가져오는 방법입니다.

```python
word = "PYTHON"

print(word[0:3])  # PYT
print(word[2:5])  # THO
print(word[:3])   # PYT
print(word[3:])   # HON
```

형식:

```python
문자열[시작:끝]
```

> 시작 위치는 포함하고, **끝 위치는 포함하지 않습니다.**

예:

```python
text = "Python"

print(text[0:3])  # Pyt
```

`0`, `1`, `2`번 글자까지만 가져옵니다.

## 7. 자주 사용하는 문자열 기능

| 기능 | 사용법 |
|---|---|
| 글자 수 | `len(text)` |
| 대문자 변경 | `text.upper()` |
| 소문자 변경 | `text.lower()` |
| 양쪽 공백 제거 | `text.strip()` |
| 문자열 변경 | `text.replace()` |
| 문자열 나누기 | `text.split()` |

```python
text = "  Python  "

print(len(text))
print(text.strip())
print(text.upper())
print(text.lower())
```

## 8. f-string

f-string은 **문자열 안에 변수 값을 넣을 때** 사용합니다.

```python
name = "찬희"
score = 90

print(f"{name}님의 점수는 {score}점입니다.")
```

계산식도 넣을 수 있습니다.

```python
score = 90

print(f"10점 추가하면 {score + 10}점입니다.")
```

## 핵심 정리
