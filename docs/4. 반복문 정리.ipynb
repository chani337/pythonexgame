# 파이썬 반복문 정리

## 1. 반복문이란?

반복문은 **같은 코드를 여러 번 실행할 때 사용하는 문법**입니다.  
쉽게 말하면 **반복해서 해야 하는 일을 자동으로 처리하는 방법**입니다.

예를 들어,

- `"안녕하세요"`를 5번 출력하기
- 1부터 10까지 출력하기
- 리스트 안의 값을 하나씩 꺼내기

같은 작업에 사용합니다.

파이썬의 대표적인 반복문에는 `for`문과 `while`문이 있습니다.

---

## 2. for문의 기본 형태

```python
for 변수 in 반복가능한값:
    실행할 코드
```

### 예시

```python
for i in range(5):
    print("안녕하세요")
```

### 실행 결과

```python
안녕하세요
안녕하세요
안녕하세요
안녕하세요
안녕하세요
```

---

## 3. range()란?

`range()`는 일정한 범위의 숫자를 만들어주는 함수입니다.

```python
for i in range(5):
    print(i)
```

### 실행 결과

```python
0
1
2
3
4
```

`range(5)`는 **0부터 4까지**를 의미합니다.

---

## 4. 원하는 범위 출력하기

```python
for i in range(1, 6):
    print(i)
```

### 실행 결과

```python
1
2
3
4
5
```

`range(1, 6)`은 **1부터 5까지**를 의미합니다.

---

## 5. 리스트와 함께 for문 사용하기

```python
fruits = ["사과", "바나나", "포도"]

for fruit in fruits:
    print(fruit)
```

### 실행 결과

```python
사과
바나나
포도
```

---

## 6. while문의 기본 형태

`while`문은 **조건이 참인 동안 계속 반복**합니다.

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

### 실행 결과

```python
0
1
2
3
4
```

---

## 7. 반복문 실습 예제 1: 1부터 10까지 출력

```python
for i in range(1, 11):
    print(i)
```

---

## 8. 반복문 실습 예제 2: 리스트 값 하나씩 출력

```python
names = ["민수", "영희", "철수"]

for name in names:
    print(name)
```

---

## 9. 반복문 실습 예제 3: 5번 반복 출력

```python
count = 0

while count < 5:
    print("파이썬 재미있다")
    count += 1
```

---

## 10. 반복문과 누적합

반복문은 값을 누적할 때도 자주 사용합니다.

```python
total = 0

for i in range(1, 6):
    total += i

print(total)
```

### 실행 결과

```python
15
```

---

## 11. 자주 하는 실수

### 1) range()의 끝 숫자도 포함된다고 생각하는 경우

```python
for i in range(1, 5):
    print(i)
```

이 코드는 1, 2, 3, 4까지만 출력됩니다.  
끝 숫자인 5는 포함되지 않습니다.

### 2) while문에서 값을 증가시키지 않는 경우

```python
i = 0

while i < 5:
    print(i)
```

이렇게 하면 `i` 값이 계속 0이므로 반복이 끝나지 않을 수 있습니다.

### 3) 들여쓰기를 틀리는 경우

```python
for i in range(3):
print(i)
```

반복문 안의 코드는 반드시 들여쓰기 해야 합니다.

---

## 12. 한 번에 정리하는 반복문 예제

```python
# 1부터 5까지 출력
for i in range(1, 6):
    print(i)

# 리스트 값 출력
fruits = ["사과", "바나나", "포도"]
for fruit in fruits:
    print(fruit)

# while문 사용
count = 0
while count < 3:
    print("반복 중")
    count += 1
```

---

## 13. 핵심 정리

- 반복문은 같은 코드를 여러 번 실행할 때 사용
- `for`문은 정해진 횟수나 데이터 개수만큼 반복할 때 사용
- `while`문은 조건이 참인 동안 반복할 때 사용
- `range()`의 끝 숫자는 포함되지 않음
- 반복문 안의 코드는 들여쓰기가 중요함
